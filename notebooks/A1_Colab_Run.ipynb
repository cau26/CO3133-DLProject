```python
import json

with open("outputs/a1_t4/linear/metrics.json", encoding="utf-8") as f:
    linear = json.load(f)

with open("outputs/a1_t4/mlp/metrics.json", encoding="utf-8") as f:
    mlp = json.load(f)

print("========== LINEAR ==========")
print(json.dumps(linear, indent=2, ensure_ascii=False))

print("\n========== MLP ==========")
print(json.dumps(mlp, indent=2, ensure_ascii=False))
```

```python
def pick(d, *keys):
    cur = d
    for k in keys:
        cur = cur[k]
    return cur

print("===== SUMMARY =====")

print("\nLinear")
print("Best epoch :", linear.get("best_epoch"))
print("Parameters :", linear.get("parameter_count"))
print("Validation :", linear.get("validation"))

print("\nMLP")
print("Best epoch :", mlp.get("best_epoch"))
print("Parameters :", mlp.get("parameter_count"))
print("Validation :", mlp.get("validation"))
```

```python
print("===== SUMMARY =====")

print("\nLinear")
print("Best epoch :", linear["best_epoch"])
print("Parameters :", linear["parameters"])
print("Val loss   :", linear["loss"])
print("Accuracy   :", linear["accuracy"])
print("Macro F1   :", linear["macro_f1"])
print("Train time :", linear["train_seconds"])
print("Infer/img  :", linear["inference"]["per_image_ms_amortized"], "ms")
print("Device     :", linear["environment"]["device"])
print("GPU        :", linear["environment"]["gpu"])

print("\nMLP")
print("Best epoch :", mlp["best_epoch"])
print("Parameters :", mlp["parameters"])
print("Val loss   :", mlp["loss"])
print("Accuracy   :", mlp["accuracy"])
print("Macro F1   :", mlp["macro_f1"])
print("Train time :", mlp["train_seconds"])
print("Infer/img  :", mlp["inference"]["per_image_ms_amortized"], "ms")
print("Device     :", mlp["environment"]["device"])
print("GPU        :", mlp["environment"]["gpu"])
```
